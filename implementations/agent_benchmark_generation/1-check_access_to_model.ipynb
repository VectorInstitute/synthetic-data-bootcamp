## Check your access to the available models

Default models for this codebase:

- `GENERATOR_MODEL=gemini-3.1-pro-preview` #gemini-3.1-pro-preview: 1.0M ctx·$2.00/$12.00 per 1M
- `AGENT_MODEL=gemini-3-flash-preview` #gemini-3-flash-preview: 1.0M ctx·$0.50/$3.00 per 1M
- `USER_SIMULATOR_MODEL=gpt-4o` #gpt-4o: 128.0K ctx·$2.50/$10.00 per 1M

You can manually set these values in `.env` or copy the variables from `.env.example` to `.env` using this command:
```bash
cp implementations/agent_benchmark_generation/.env.example .env
```


````python
import os

from dotenv import load_dotenv
from openai import OpenAI


# Loads your environment variables from the .env file
load_dotenv()
client = OpenAI(base_url=os.getenv("SYNBENCH_BASE_URL"), api_key=os.environ.get("OPENAI_API_KEY"))

stream = client.chat.completions.create(
    model="gemini-3.1-flash-lite-preview",
    messages=[{"role": "user", "content": "Hi, how are you?"}],
    stream=True,
)

for chunk in stream:
    if chunk.choices:
        print(chunk.choices[0].delta.content or "", end="\n", flush=True)


print("<--DEFAULT MODELS-->")
print(f"GENERATOR_MODEL{os.getenv('GENERATOR_MODEL')}")
print(f"AGENT_MODEL{os.getenv('AGENT_MODEL')}")
print(f"USER_SIMULATOR_MODEL{os.getenv('USER_SIMULATOR_MODEL')}")
````
